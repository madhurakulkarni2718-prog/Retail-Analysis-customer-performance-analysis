```python
# Day 25 — Sales EDA

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("transformed_retail_sales.csv")

df.head()
```

```python
# Task 2 — Monthly Sales

monthly_sales = df.groupby("Month_Name")["Sales"].sum()

print(monthly_sales)
```

```python
# Task 3 — Category Sales

category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
# Category Sales Bar Chart

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 4 — Regional Sales

regional_sales = df.groupby("Region")["Sales"].sum()

print(regional_sales)
```

```python
# Regional Sales Bar Chart

regional_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 5 — Top 10 Products

top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
# Bottom 10 Products

bottom_products = df.groupby("Product Name")["Sales"].sum().sort_values(
    ascending=True
).head(10)

print(bottom_products)
```

```python
# Task 6 — Total Profit

total_profit = df["Profit"].sum()

print("Total Profit:", total_profit)
```

```python
# Profit by Category

profit_category = df.groupby("Category")["Profit"].sum()

print(profit_category)
```

```python
# Profit by Region

profit_region = df.groupby("Region")["Profit"].sum()

print(profit_region)
```

```python
# Profit by Product

profit_product = df.groupby("Product Name")["Profit"].sum().sort_values(ascending=False)

print(profit_product.head(10))
```

```python
# Profit by Category Chart

profit_category.plot(kind="bar")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# Profit by Region Chart

profit_region.plot(kind="bar")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# Sales Distribution

plt.figure(figsize=(8, 5))

sns.histplot(df["Sales"], bins=30, kde=True)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Top 10 Products Chart

top_products.plot(kind="bar")

plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.show()
```

```python
# Save Sales EDA Data

df.to_csv("sales_eda_data.csv", index=False)

print("File saved successfully!")
```
