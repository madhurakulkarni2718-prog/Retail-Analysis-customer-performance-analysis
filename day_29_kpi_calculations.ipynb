```python
# DAY 29 — KPI DEVELOPMENT

import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")

print("Dataset Loaded Successfully!")
display(df.head())
```

```python
# KPI 1: Total Revenue

total_revenue = df["Sales"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python
# KPI 2: Total Profit

total_profit = df["Profit"].sum()

print("Total Profit:", round(total_profit, 2))
```

```python
# KPI 2: Total Profit

total_profit = df["Profit"].sum()

print("Total Profit:", round(total_profit, 2))
```

```python
# KPI 3: Total Orders

total_orders = df["Order ID"].nunique()

print("Total Orders:", total_orders)
```

```python
# KPI 4: Total Customers

total_customers = df["Customer ID"].nunique()

print("Total Customers:", total_customers)
```

```python
# KPI 5: Average Order Value

aov = total_revenue / total_orders if total_orders > 0 else 0

print("Average Order Value:", round(aov, 2))
```

```python
# KPI 6: Profit Margin

profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue != 0 else 0
)

print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# KPI 7: Growth Percentage

df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
    .sort_index()
)

if len(monthly_sales) >= 2 and monthly_sales.iloc[-2] != 0:
    growth_percentage = (
        (monthly_sales.iloc[-1] - monthly_sales.iloc[-2])
        / monthly_sales.iloc[-2]
    ) * 100

    print("Growth Percentage:", round(growth_percentage, 2), "%")
else:
    growth_percentage = None
    print("Growth Percentage: Not Available")
```

```python
# Display KPI Summary

kpi_summary = pd.DataFrame({
    "KPI Name": [
        "Total Revenue",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Average Order Value",
        "Profit Margin (%)",
        "Growth (%)"
    ],
    "Value": [
        round(total_revenue, 2),
        round(total_profit, 2),
        total_orders,
        total_customers,
        round(aov, 2),
        round(profit_margin, 2),
        round(growth_percentage, 2)
        if growth_percentage is not None else "Not Available"
    ]
})

display(kpi_summary)
```

```python
# Create Day 29 KPI Document

from docx import Document

doc = Document()

doc.add_heading("Day 29 — KPI Development", 0)
doc.add_paragraph("Dataset: Sample - Superstore.csv")
doc.add_paragraph(
    "Objective: Convert raw analysis into management-level KPIs."
)

kpis = [
    (
        "Total Revenue",
        "Total sales revenue generated.",
        "SUM(Sales)",
        "Shows the total value of sales.",
        "Management uses it to track business revenue."
    ),
    (
        "Total Profit",
        "Total profit generated.",
        "SUM(Profit)",
        "Shows the overall profit or loss.",
        "Management uses it to evaluate profitability."
    ),
    (
        "Total Orders",
        "Count of unique orders.",
        "COUNT DISTINCT Order ID",
        "Shows the number of different orders.",
        "Management uses it to monitor order volume."
    ),
    (
        "Total Customers",
        "Count of unique customers.",
        "COUNT DISTINCT Customer ID",
        "Shows how many different customers purchased.",
        "Management uses it to understand customer reach."
    ),
    (
        "Average Order Value",
        "Average revenue per unique order.",
        "Total Revenue / Total Orders",
        "Shows the average value of an order.",
        "Management uses it to monitor customer spending per order."
    ),
    (
        "Profit Margin",
        "Profit as a percentage of revenue.",
        "Total Profit / Total Revenue * 100",
        "Shows how much profit is earned per 100 units of revenue.",
        "Management uses it to assess profitability."
    ),
    (
        "Growth Percentage",
        "Percentage change in revenue between periods.",
        "(Current Revenue - Previous Revenue) / Previous Revenue * 100",
        "Shows whether revenue increased or decreased.",
        "Management uses it to monitor business growth."
    )
]

for name, definition, formula, meaning, reason in kpis:
    doc.add_heading(name, level=1)
    doc.add_paragraph("Definition: " + definition)
    doc.add_paragraph("Formula: " + formula)
    doc.add_paragraph("Business Meaning: " + meaning)
    doc.add_paragraph("Why Management Needs It: " + reason)

doc.add_heading("Calculated KPI Results", level=1)

table = doc.add_table(rows=1, cols=2)
table.style = "Table Grid"
table.rows[0].cells[0].text = "KPI Name"
table.rows[0].cells[1].text = "Value"

for _, row in kpi_summary.iterrows():
    cells = table.add_row().cells
    cells[0].text = str(row["KPI Name"])
    cells[1].text = str(row["Value"])

doc.add_heading("Conclusion", level=1)
doc.add_paragraph(
    "These KPIs summarize revenue, profit, order volume, customer count, "
    "average order value, profit margin and revenue growth for management."
)

doc.save("Day_29_KPI_Document.docx")

print("Day_29_KPI_Document.docx created successfully!")
```
