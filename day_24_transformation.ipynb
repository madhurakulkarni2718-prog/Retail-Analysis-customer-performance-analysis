```python
import pandas as pd

df = pd.read_csv("cleaned_data.csv")

df.head()
```

```python
print(df.columns.tolist())
```

```python
# Task 1 — Date Features

df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
# Create Date Features

df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month_Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter

df.head()
```

```python
# Task 2 — Profit Margin

df["Profit_Margin"] = (
    df["Profit"] / df["Sales"]
) * 100

df.head()
```

```python
# Task 3 — Order Value

df["Order_Value"] = df["Sales"]

df.head()
```

```python
# Task 4 — Customer Segmentation

df["Customer_Segment"] = pd.cut(
    df["Sales"],
    bins=[-float("inf"), 1000, 5000, float("inf")],
    labels=["Low Value", "Medium Value", "High Value"]
)

df.head()
```

```python
df["Customer_Segment"] = pd.cut(
    df["Sales"],
    bins=[-float("inf"), 1000, 5000, float("inf")],
    labels=["Low Value", "Medium Value", "High Value"]
)

df[["Sales", "Customer_Segment"]].head()
```

```python
# Task 5 — Additional Analytical Columns

df["Sales_Category"] = df["Sales"].apply(
    lambda x: "High" if x > 5000 else
    "Medium" if x > 1000 else "Low"
)

df["Profit_Category"] = df["Profit"].apply(
    lambda x: "Profit" if x > 0 else "Loss"
)

df["Customer_Type"] = df["Segment"]

df["Product_Performance"] = df["Profit"].apply(
    lambda x: "Good" if x > 0 else "Needs Improvement"
)

df["Order_Size"] = df["Quantity"].apply(
    lambda x: "Large" if x >= 5 else "Small"
)

df.head()
```

```python
# Save Transformed Data

df.to_csv("transformed_retail_sales.csv", index=False)

print("File saved successfully!")
```
